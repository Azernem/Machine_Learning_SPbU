```python
import networkx as nx
import numpy as np
import torch

def get_dataset_graph():
    # Генерируем списки графов
    star_graphs = [nx.star_graph(np.random.randint(4, 13)) for i in range(100)]
    complete_graphs = [nx.complete_graph(np.random.randint(4, 13)) for i in range(100)]

    # Объединяем списки (создаем новый список)
    all_graphs = star_graphs + complete_graphs

    # Переводим каждый граф в матрицу смежности (массив numpy)
    # Чтобы тензор был ровным, приводим все к одному размеру (например, 13x13)
    adj_matrices = []
    for G in all_graphs:
        adj = nx.to_numpy_array(G, nodelist=range(13), weight=None) # weight=None делает 0 и 1
        adj_matrices.append(adj)

    # Создаем тензор: размер будет [200, 13, 13]
    dataset = torch.tensor(np.array(adj_matrices), dtype=torch.float32)

    return dataset

# Если нужны еще и метки (0 для звезд, 1 для полных)
labels = torch.cat([torch.zeros(100), torch.ones(100)]).long()
labels
```
